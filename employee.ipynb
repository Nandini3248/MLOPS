```python
import pandas as pd

d = pd.read_csv(r"C:\Users\Student\Downloads\employees.csv")
print(d)
```

```python
removed_data=pd.DataFrame(columns=d.columns)
original_data=pd.DataFrame(columns=d.columns)
for index, row in d.iterrows():
    found=False
    for column in d.columns:
        if pd.isnull(row[column]):
            found=True
            break
    if found:
        removed_data.loc[len(removed_data)]=row
    else:
        original_data.loc[len(original_data)]=row
```
